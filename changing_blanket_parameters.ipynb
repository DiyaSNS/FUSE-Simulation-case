```julia
# See blanket specific layers from build
for layer in dd.build.layer
    if occursin("blanket", lowercase(string(layer.name))) || 
       occursin("wall", lowercase(string(layer.name))) ||
       occursin("shield", lowercase(string(layer.name)))
        println("Name: ", layer.name)
        println("Material: ", layer.material)
        println("Thickness: ", layer.thickness, " m")
        println("Volume: ", layer.volume, " m³")
        println("---")
    end
end
```

```julia
# See all act.ActorBlanket parameters with values
for param in propertynames(act.ActorBlanket)
    if param != :_parent && param != :_name && param != :_time
        try
            println(param, " = ", getproperty(act.ActorBlanket, param))
        catch
        end
    end
end
```

```julia
ini, act = FUSE.case_parameters(:ITER; init_from=:scalars)

# realistic thermal efficiency value
act.ActorBlanket.thermal_power_extraction_efficiency = 0.88

# Changing neutron multiplier (typical range 1.1 - 1.3)
act.ActorBlanket.blanket_multiplier = 1.27

# Changing Li6 enrichment (natural Li has ~7.5% Li6, enriched up to 90%)
act.ActorBlanket.max_Li6_enrichment_fraction = 0.6

# Changing minimum first wall thickness
act.ActorBlanket.minimum_first_wall_thickness = 0.03

dd = IMAS.dd()
FUSE.init(dd, ini, act)
FUSE.ActorWholeFacility(dd, act)

# Run blanket actor explicitly
FUSE.ActorBlanket(dd, act)

# Check blanket results
println(dd.blanket)
```
